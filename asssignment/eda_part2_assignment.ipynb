# Step 4: Feature Relationships

In this step, we will explore the relationships between different features in the dataset using scatterplots, heatmaps, and pairplots to identify potential correlations.

* Scatterplot
* Heatmap Correlation
* Pairplot
* Groupby comparisons

```python
df.plot(kind='scatter',
        x='Speed_mph',
        y='Height_ft',
        title='Coaster Speed vs. Height')
plt.show()
```

```python
ax = sns.scatterplot(x='Speed_mph',
                y='Height_ft',
                hue='Year_Introduced',
                data=df)
ax.set_title('Coaster Speed vs. Height')
plt.show()
```

```python
sns.pairplot(df,
             vars=['Year_Introduced','Speed_mph',
                   'Height_ft','Inversions','Gforce'],
            hue='Type_Main')
plt.show()
```

```python
df_corr = df[['Year_Introduced','Speed_mph',
    'Height_ft','Inversions','Gforce']].dropna().corr()
df_corr
```

```python
sns.heatmap(df_corr, annot=True)
```
